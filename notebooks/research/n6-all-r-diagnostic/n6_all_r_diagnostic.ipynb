```python
paths=pd.read_csv(OUT/'capital_paths.csv')
fig,ax=plt.subplots(figsize=(10,5))
for arm,g in paths[paths.scope.eq('ALL')].groupby('arm'):
    ax.plot(g.issue_order,g.capital_cny,label=arm)
ax.axhline(100000,color='black',lw=.8); ax.legend(title='arm'); ax.set_title('Corrected N6 fixed-R capital paths (ALL)'); plt.show()
```

```python
lift=contrasts[contrasts.arm.ne('R0')]
fig,ax=plt.subplots(figsize=(10,4))
for scope,g in lift.groupby('scope'):
    ax.plot(g.arm,g.matched_lift_cny,marker='o',label=scope)
ax.axhline(0,color='black',lw=.8); ax.legend(); ax.set_title('Matched lift vs corrected R0'); plt.show()
```
